## #3

```python
#Genera matrices de números aleatorios utilizando np.random.rand() (valores entre 0 y 1) y np.random.randint() (valores enteros).
rand_float = np.random.rand(3, 3)
rand_int = np.random.randint(0, 100, (3, 3))
print(rand_float)
print(rand_int)
```

## #4

```python
#Construye matrices estructurales empleando np.zeros() , np.ones() y una matriz identidad con np.eye() .
print('zeros:\n', np.zeros((3, 3)))
print('ones:\n', np.ones((3, 3)))
print('eye:\n', np.eye(3))
```

## #5

```python
#Traspone una matriz generada aleatoriamente utilizando el atributo .T .
m = np.random.randint(0, 10, (2, 3))
print('Original:\n', m, m.shape)
print('Transpuesta:\n', m.T, m.T.shape)
```

## #6

```python
#Concatena dos arreglos de manera horizontal ( np.hstack() ) y de manera vertical ( np.vstack() ).
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print('hstack:\n', np.hstack((a, b)))
print('vstack:\n', np.vstack((a, b)))
```

## #7

```python
#Ejecuta operaciones matemáticas sobre los arreglos para calcular: suma ( np.sum() ), promedio
#( np.mean() ), mediana ( np.median() ), producto ( np.prod() ), desviación estándar
#( np.std() ), varianza ( np.var() ), valores mínimos y máximos, y la suma acumulativa ( np.cumsum() ).
datos = np.array([4, 8, 15, 16, 23, 42])
print('Suma:', np.sum(datos))
print('Promedio:', np.mean(datos))
print('Mediana:', np.median(datos))
print('Producto:', np.prod(datos))
print('Desv. estándar:', np.std(datos))
print('Varianza:', np.var(datos))
print('Mínimo:', np.min(datos))
print('Máximo:', np.max(datos))
print('Suma acumulativa:', np.cumsum(datos))
```

## #8

```python
#Demuestra las operaciones aritméticas elemento a elemento (suma, resta y multiplicación) entre arreglos.
x = np.array([10, 20, 30])
y = np.array([1, 2, 3])
print('Suma:', x + y)
print('Resta:', x - y)
print('Multiplicación:', x * y)
```
